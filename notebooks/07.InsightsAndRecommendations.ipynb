5. Name 3 outlier removal methods?
    - **Z-Score Method**: Identifies outliers by measuring how many standard deviations a data point is from the mean.
    - **IQR Method**: Uses the interquartile range to determine outliers by identifying points that fall below Q1 - 1.5*IQR or above Q3 + 1.5*IQR.
    - **Isolation Forest**: An algorithm that isolates outliers instead of profiling normal data points.
